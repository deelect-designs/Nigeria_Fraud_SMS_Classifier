```python
import warnings
warnings.filterwarnings("ignore")

from pathlib import Path

import joblib
import pandas as pd

from sklearn.feature_extraction.text import TfidfVectorizer

print("Libraries imported successfully.")
```

```python
CURRENT_DIR = Path.cwd()

if CURRENT_DIR.name == "notebooks":
    PROJECT_ROOT = CURRENT_DIR.parent
else:
    PROJECT_ROOT = CURRENT_DIR

DATA_PATH = PROJECT_ROOT / "data" / "spam_processed.csv"

df = pd.read_csv(DATA_PATH)

print(df.head())
```

```python
print("="*50)
print(df.info())
print("="*50)

print(df.isnull().sum())
```

```python
X = df["processed_message"]

y = df["target"]

print("Number of samples:", len(X))
print("Number of labels :", len(y))
```

```python
tfidf = TfidfVectorizer(
    lowercase=False,
    stop_words=None,
    max_features=5000,
    ngram_range=(1,2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

print("TF-IDF Vectorizer created.")
```

```python
X_tfidf = tfidf.fit_transform(X)

print("TF-IDF transformation completed.")
```

```python
print("Shape of feature matrix:")
print(X_tfidf.shape)
```

```python
print("Vocabulary Size:", len(tfidf.vocabulary_))
```

```python
feature_names = tfidf.get_feature_names_out()

print(feature_names[:50])
```

```python
feature_df = pd.DataFrame(
    X_tfidf[:5].toarray(),
    columns=tfidf.get_feature_names_out()
)

feature_df.head()
```

```python
import numpy as np

importance = np.asarray(X_tfidf.sum(axis=0)).flatten()

top_features = pd.DataFrame({
    "Feature": tfidf.get_feature_names_out(),
    "Score": importance
})

top_features = top_features.sort_values(
    by="Score",
    ascending=False
)

top_features.head(20)
```

```python
MODELS_DIR = PROJECT_ROOT / "models"

MODELS_DIR.mkdir(exist_ok=True)

joblib.dump(
    tfidf,
    MODELS_DIR / "tfidf.pkl"
)

print("TF-IDF Vectorizer saved.")
```

```python
joblib.dump(
    X_tfidf,
    MODELS_DIR / "X_tfidf.pkl"
)

joblib.dump(
    y,
    MODELS_DIR / "y.pkl"
)

print("Feature matrix saved.")
```

```python
sample = [
    "Congratulations you have won one million naira claim now"
]

sample_vector = tfidf.transform(sample)

print(sample_vector.shape)
```

```python
non_zero = sample_vector.nonzero()[1]

features = tfidf.get_feature_names_out()

for index in non_zero:
    print(features[index])
```

```python
processed = df[
    [
        "processed_message",
        "target"
    ]
]

processed.to_csv(
    PROJECT_ROOT / "data" / "training_dataset.csv",
    index=False
)

print("Training dataset saved.")
```
